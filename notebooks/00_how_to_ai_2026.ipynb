## 1. UV - Python Package Manager

[UV](https://docs.astral.sh/uv/) เป็น Python package manager ตัวใหม่ที่เร็วมาก และสามารถทดแทน pip, poetry, pyenv, virtualenv และเครื่องมืออื่นๆ ได้

### การติดตั้ง UV

บน macOS/Linux:
```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

บน Windows (PowerShell):
```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

### การใช้งาน UV

สร้างโปรเจคใหม่ (ระบุชื่อโปรเจคและเวอร์ชัน Python):
```bash
uv init <project_name> --python 3.12
cd <project_name>
```

ตัวอย่างเช่น:
```bash
uv init my-ai-project --python 3.12
cd my-ai-project
```

flag `--python 3.12` จะกำหนดให้โปรเจคใช้ Python 3.12 ซึ่งเป็นเวอร์ชันที่แนะนำสำหรับ AI Builders 2026

คำสั่ง `uv init` จะสร้างโครงสร้างโปรเจคให้อัตโนมัติ รวมถึง:
- `pyproject.toml` - ไฟล์ config สำหรับโปรเจค (ระบุ Python version, dependencies)
- `.python-version` - ระบุเวอร์ชัน Python (จะเป็น 3.12 ตามที่กำหนด)
- `.venv/` - virtual environment
- `main.py` - ไฟล์ Python หลัก
- Initialize git repository

ติดตั้ง package:
```bash
uv add numpy pandas
```

รัน Python:
```bash
uv run python main.py
```

[LM Studio](https://lmstudio.ai/) เป็น application สำหรับรัน LLM บนเครื่อง local ของเรา สามารถใช้เป็น chat app และมี API server ให้ด้วย


## 6. Hugging Face

[Hugging Face](https://huggingface.co/) เป็น platform หลักสำหรับ AI/ML มีทั้ง model repository, datasets, และ **Inference Engine** ที่สำคัญมาก

### คุณสมบัติ

- **Model Hub** - คลังโมเดล AI ที่ใหญ่ที่สุด
- **Datasets** - ชุดข้อมูลสำหรับเทรนและทดสอบ
- **Spaces** - hosting สำหรับ demo applications
- **Inference Providers** - เรียกใช้โมเดลผ่าน API ได้ทันที โดยไม่ต้อง deploy เอง

### HF Inference Engine (สำคัญมาก!)

HF Inference Engine คือ unified interface สำหรับเรียกใช้โมเดลบน Hugging Face Hub ผ่าน `InferenceClient` สามารถเชื่อมต่อกับ provider หลายเจ้า (Replicate, Together AI, Cerebras, Nebius, Groq, ฯลฯ) หรือแม้แต่ local server ได้

ข้อดีหลัก:
- **ใช้ได้ทันที** - ไม่ต้อง deploy server เอง
- **OpenAI-compatible** - เปลี่ยนจาก OpenAI มาใช้ได้ง่ายมาก แค่เปลี่ยน 2 บรรทัด
- **หลาย providers** - เลือก provider ที่ถูกที่สุดหรือเร็วที่สุดได้
- **หลาย tasks** - รองรับ chat completion, text-to-image, speech recognition, ฯลฯ

### การติดตั้ง
```bash
uv add huggingface_hub
```

### ตัวอย่าง: Chat Completion ผ่าน HF Inference
```python
from huggingface_hub import InferenceClient

# สร้าง client (ใช้ HF token สำหรับ authentication)
client = InferenceClient(
    provider="auto",  # เลือก provider อัตโนมัติ หรือระบุเช่น "together", "nebius"
)

# เรียกใช้ LLM
response = client.chat.completions.create(
    model="meta-llama/Llama-3.1-8B-Instruct",
    messages=[{"role": "user", "content": "อธิบาย Machine Learning แบบง่ายๆ"}],
    max_tokens=200,
)
print(response.choices[0].message.content)
```

### ตัวอย่าง: เปลี่ยนจาก OpenAI มาใช้ HF (แค่ 2 บรรทัด!)
```python
# เดิม (OpenAI)
# from openai import OpenAI
# client = OpenAI(api_key="sk-...")

# ใหม่ (Hugging Face) - แค่เปลี่ยน import และ client!
from huggingface_hub import InferenceClient
client = InferenceClient()  # ใช้ HF token ที่ login ไว้แล้ว

# code ที่เหลือเหมือนเดิมทุกอย่าง!
output = client.chat.completions.create(
    model="meta-llama/Llama-3.1-8B-Instruct",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Hello!"},
    ],
    stream=True,
    max_tokens=512,
)

for chunk in output:
    print(chunk.choices[0].delta.content, end="")
```

### ตัวอย่าง: Text-to-Image
```python
from huggingface_hub import InferenceClient

client = InferenceClient(provider="fal-ai")
image = client.text_to_image(
    "A cute robot learning AI in a classroom",
    model="black-forest-labs/FLUX.1-schnell",
)
image.save("robot_student.png")
```

### ตัวอย่าง: เชื่อมต่อกับ Local Server (LM Studio, Ollama, etc.)
```python
from huggingface_hub import InferenceClient

# เชื่อมต่อกับ LM Studio หรือ local server ที่รัน OpenAI-compatible API
client = InferenceClient(model="http://localhost:1234")

response = client.chat.completions.create(
    messages=[{"role": "user", "content": "สวัสดี!"}],
    max_tokens=100,
)
print(response.choices[0].message.content)
```

### OpenCode - แนะนำสำหรับ AI Builders

[OpenCode](https://opencode.ai/) เป็น open source AI coding agent ที่ทำงานบน terminal ใช้งานง่าย ฟรี และทรงพลังมาก

#### การติดตั้ง OpenCode

วิธีที่ง่ายที่สุด:
```bash
curl -fsSL https://opencode.ai/install | bash
```

หรือผ่าน npm:
```bash
npm install -g opencode-ai
```

หรือผ่าน Homebrew (macOS/Linux):
```bash
brew install anomalyco/tap/opencode
```

#### การตั้งค่า OpenCode

OpenCode รองรับ LLM providers หลายเจ้า วิธีที่ง่ายที่สุดคือใช้ **OpenCode Zen**:

1. รัน OpenCode ในโปรเจค:
```bash
cd my-ai-project
opencode
```

2. ใช้คำสั่ง `/connect` เพื่อเชื่อมต่อกับ provider:
```
/connect
```

3. เลือก provider (เช่น opencode, anthropic, openai) แล้วใส่ API key

#### การเริ่มต้นใช้งานในโปรเจค

เมื่อเปิด OpenCode ในโปรเจค ให้รัน `/init` เพื่อให้ OpenCode วิเคราะห์โปรเจค:
```
/init
```

คำสั่งนี้จะสร้างไฟล์ `AGENTS.md` ที่ช่วยให้ OpenCode เข้าใจโครงสร้างและ patterns ของโปรเจค (ควร commit ไฟล์นี้เข้า Git ด้วย)

#### ตัวอย่างการใช้งาน OpenCode

**1. ถาม OpenCode เกี่ยวกับ codebase:**
```
How is authentication handled in @packages/functions/src/api/index.ts
```
ใช้ `@` เพื่อ fuzzy search หาไฟล์ในโปรเจค

**2. วางแผนก่อนเขียน code (Plan mode):**

กด `Tab` เพื่อสลับเป็น Plan mode (จะเห็น indicator มุมขวาล่าง)
```
สร้างระบบ login ด้วย email และ password พร้อม validation
```
OpenCode จะวางแผนให้ก่อน แล้วค่อยกด `Tab` กลับมา Build mode เพื่อสร้างจริง

**3. สร้าง feature ใหม่ (Build mode):**
```
สร้าง REST API endpoint สำหรับ CRUD operations ของ users
พร้อม validation และ error handling
```

**4. แก้ bug:**
```
มี error TypeError: Cannot read property 'map' of undefined
ใน src/components/UserList.tsx ช่วยดูและแก้ให้หน่อย
```

**5. Refactor code:**
```
Refactor ฟังก์ชันใน @src/utils/helpers.py ให้ clean ขึ้น
แยกฟังก์ชันที่ยาวเกินไปออกเป็นฟังก์ชันย่อย
```

**6. เขียน tests:**
```
เขียน unit tests สำหรับ @src/services/auth.py ให้ครอบคลุม edge cases
```

**7. อธิบาย code:**
```
อธิบายว่า @src/models/transformer.py ทำงานอย่างไร step by step
```

#### คำสั่งสำคัญใน OpenCode

| คำสั่ง | การทำงาน |
|--------|----------|
| `/init` | วิเคราะห์โปรเจคและสร้าง AGENTS.md |
| `/connect` | เชื่อมต่อ/เปลี่ยน LLM provider |
| `/undo` | ย้อนกลับการเปลี่ยนแปลงล่าสุด |
| `/redo` | ทำซ้ำการเปลี่ยนแปลงที่ undo |
| `/share` | แชร์ conversation เป็น link |
| `Tab` | สลับระหว่าง Plan mode / Build mode |
| `@` | Fuzzy search หาไฟล์ในโปรเจค |

#### ทำไมต้อง OpenCode?

- **Open source** และ **ฟรี** - เลือกใช้ LLM provider ที่ต้องการได้
- **ทำงานบน terminal** - เบา เร็ว ไม่ต้องเปิด IDE หนักๆ
- **อ่าน/เขียน/รัน code ได้** - ไม่ใช่แค่ autocomplete แต่เข้าใจทั้งโปรเจค
- **Plan + Build mode** - วางแผนก่อนเขียนจริง ลดความผิดพลาด
- **Undo/Redo** - ย้อนกลับได้ถ้าไม่พอใจ
- **รองรับ drag & drop รูปภาพ** - ส่งรูป design ให้ OpenCode ดูได้
- **Share conversations** - แชร์การสนทนากับทีมได้

### คำแนะนำในการใช้ Coding Agent

- ควรใช้ coding agent เป็นเครื่องมือช่วย แต่ต้องเข้าใจ code ที่มันสร้างด้วย

## 8. Docker (Optional)
